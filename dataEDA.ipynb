```python
def convert_range_to_mean(value):
    if pd.isna(value):
        return value
    elif '–' in str(value):  
        start, end = map(int, value.split('–'))
        return int((start + end) / 2) 
    elif '+' in str(value):  
        return int(value.replace('+', ''))  
    return value

data['price_per_person'] = data['price_per_person'].apply(convert_range_to_mean)

print(data[['rating', 'price_per_person']].head(50))
```

```python
def replace_nan_based_on_rating(row):
    if pd.isna(row['price_per_person']):
        if row['rating'] >= 4.6:
            return 1000
        elif row['rating'] >= 4.5:
            return 750
        elif row['rating'] >= 4.2:
            return 650
        else:
            return 450
    return row['price_per_person']

data['price_per_person'] = data.apply(replace_nan_based_on_rating, axis=1)

print(data[['rating', 'price_per_person']].head(50))
```

```python
def replace_review_count(review_text):
    if isinstance(review_text, str):  
        return re.sub(r'Review (\d+) \(\d+ stars\):', r'\1:', review_text)
    return review_text  


data['sample_user_reviews'] = data['sample_user_reviews'].apply(replace_review_count)


print(data['sample_user_reviews'][2])
```

```python
def remove_emojis(text):
    if isinstance(text, str):  
        emoji_pattern = re.compile(
            "[\U00010000-\U0010ffff"  
            "\u200d"                 
            "\u20e3"                  
            "\u2194-\u21AA"          
            "\u2300-\u23FF"           
            "\u2600-\u26FF"           
            "\u2700-\u27BF"          
            "\u2B50"                 
            "]+", flags=re.UNICODE)
        return emoji_pattern.sub(r'', text)
    return text  
data['sample_user_reviews'] = data['sample_user_reviews'].apply(remove_emojis)
```

```python
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.preprocessing import StandardScaler
import mysql.connector

# Connect to MySQL database
db_connection = mysql.connector.connect(
    host="localhost", 
    user="mukul", 
    password="mukul1572", 
    database="feast_ai"
)

# Create a cursor object
cursor = db_connection.cursor()

# Query to fetch data from 'restaurants' table
query = "SELECT * FROM restaurants;"
df = pd.read_sql(query, db_connection)

# Initialize the pre-trained model for generating embeddings
model = SentenceTransformer('all-MiniLM-L6-v2')
scaler = StandardScaler()

# Function to generate embeddings
def create_embeddings(df):
    # Prepare text fields (name, address, sample_user_reviews)
    text_fields = df[['name', 'address', 'sample_user_reviews']].fillna('')
    text_data = text_fields.apply(lambda row: ' '.join(row), axis=1)
    
    # Generate embeddings for the text fields
    text_embeddings = model.encode(text_data.tolist())

    # Handle numerical fields (rating, reviews_count, latitude, longitude, price_per_person)
    numerical_fields = df[['rating', 'reviews_count', 'latitude', 'longitude', 'price_per_person']].fillna(0)
    numerical_embeddings = scaler.fit_transform(numerical_fields)

    # Combine both textual and numerical embeddings
    combined_embeddings = [list(text_embeddings[i]) + list(numerical_embeddings[i]) for i in range(len(df))]
    return combined_embeddings

# Generate the embeddings
embeddings = create_embeddings(df)

# Add the embeddings to the dataframe
df['embedding'] = embeddings

# Insert the embeddings into the MySQL database (as string)
for idx, row in df.iterrows():
    embedding_str = str(row['embedding'])  # Convert list to string
    cursor.execute(f"UPDATE restaurants SET embedding = %s WHERE id = %s", (embedding_str, row['id']))

# Commit the changes to the database
db_connection.commit()

# Optionally, you can print the results for verification
cursor.execute("SELECT id, name, embedding FROM restaurants LIMIT 10;")
result = cursor.fetchall()

# Print the first 10 rows to check the embeddings
for row in result:
    print(row)

# Close the cursor and connection
cursor.close()
db_connection.close()
```
